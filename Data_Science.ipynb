```python
#13.Sayısal eksik verilerin ortalama ile doldurulması
df_average=df.copy()
numeric_column=df_average.select_dtypes(include=["int64","float64"]).columns
numeric_column
```

```python
for column in numeric_column:
  df_average[column]=df_average[column].fillna(df_average[column].mean())
df_average[numeric_column].isnull().sum()
```

```python
#14.Sayısal eksik verileri medyan ile doldurma
df_median=df.copy()
for column in numeric_column:
  df_median[column]=df_median[column].fillna(df_median[column].median())
df_median[numeric_column].isnull().sum()
```

```python
#15.Sayısal eksik verileri mod ile doldurma
df_mod=df.copy()
for  column in numeric_column:
  df_mod[column]=df_mod[column].fillna(df_mod[column].mode()[0])
df_mod[numeric_column].isnull().sum()
```

```python
#16.Kategorik eksik verileri mod ile doldurma
df_mod_categorical=df.copy()
categorical_column=df_mod_categorical.select_dtypes(include=["object"]).columns
categorical_column
```

```python
for column in categorical_column:
  mod_value=df_mod_categorical[column].mode()[0]
  df_mod_categorical[column]=df_mod_categorical[column].fillna(mod_value)
df_mod_categorical[categorical_column].isnull().sum()
```

```python
#17.Kategorik eksik verileri sabit bir değer ile doldurulalım
df_constant_categorical=df.copy()
for column in categorical_column:
  df_constant_categorical[column]=df_constant_categorical[column].fillna("Bilinmiyor")
df_constant_categorical[categorical_column].isnull().sum()
```

```python
#18.Eğitimde kullanacağımız güncel veri seti
#sayısal eksik verileri medyan ile doldur
#Kategorik eksik verileri mod ile doldur
df_current=df.copy()
for column in numeric_column:
  df_current[column]=df_current[column].fillna(df_current[column].median())
for column in categorical_column:
  mod_value=df_current[column].mode()[0]
  df_current[column]=df_current[column].fillna(mod_value)
```

```python
#33.Şehir sütunundaki tutarsız değerleri düzeltme
def normalize(series):
  return(
      series.astype(str)
      .str.strip()
      .str.lower()
      .str.replace("i̇","i",regex=False)
      .str.replace("ı","i",regex=False)
      .str.replace("ş","s",regex=False)
      .str.replace("ç","c",regex=False)
      .str.replace("ö","o",regex=False)
      .str.replace("ğ","g",regex=False)
  )
```

```python
df["sehir"]=normalize(df["sehir"])
df["sehir"].unique()
```

```python
sehir_duzeltme_haritasi = {
    "istanbul": "İstanbul",
    "ankara": "Ankara",
    "izmir": "İzmir",
    "antalya": "Antalya",
    "konya": "Konya",
    "bursa": "Bursa",
    "adana": "Adana",
    "gaziantep": "Gaziantep",
    "mersin": "Mersin",
    "kayseri": "Kayseri",
    "eskisehir": "Eskişehir",
    "samsun": "Samsun",
    "trabzon": "Trabzon",
    "kocaeli": "Kocaeli"
}
df["sehir"]=df["sehir"].replace(sehir_duzeltme_haritasi)
df["sehir"].unique()
```

```python
#40.Veri setini yükle
df=pd.read_csv("e_ticaret_veri_seti_tekrarlayan_kayıtlar_duzenlendi.csv")
```

```python
#41.Sayısal sütunların belirlenmesi
numeric_column=df.select_dtypes(include=["int64","float64"]).columns.tolist()
numeric_column
```

```python
#42.Sayısal sütunların özet istatisliklerini incelemek
df[numeric_column].describe()
```

```python
"""
IQR=Q3-Q1=817-198=619 For birim_fiyat
Üst sınır=Q3+k*IQR
Alt sınır=Q1-k*IQR
k=1 olsun
Üst sınır=817+619=1436
ALt sınır=198-619=-421
Outlier=üst sınır ve alt sınır dışında kalan örnekler
"""
```

```python
#43.IQR ile aykırı değer inceleme
aykiri_degerler={}

for column in numeric_column:
  q1=df[column].quantile(0.25)
  q3=df[column].quantile(0.75)
  iqr=q3-q1
  ust_sinir=q3+1.5*iqr
  alt_sinir=q1-1.5*iqr

  aykiri_degerler[column]={
      "q1":q1,
      "q3":q3,
      "iqr":iqr,
      "ust_sinir":ust_sinir,
      "alt_sinir":alt_sinir
  }
```

```python
aykiri_degerler
```

```python
#44.Aykırı değer içeren kayıtları inceleyelim
aykiri_maskesi=pd.Series(False,index=df.index)
for column in numeric_column:
  alt_sinir=aykiri_degerler[column]["alt_sinir"]
  ust_sinir=aykiri_degerler[column]["ust_sinir"]
  sutun_maskesi=(df[column]<alt_sinir)|(df[column]>ust_sinir)
  aykiri_maskesi=aykiri_maskesi|sutun_maskesi
aykiri_kayıtlar=df[aykiri_maskesi]
aykiri_kayıtlar
```

```python
#45.Aykırı değer sayılarını sütun bazında incele
for column in numeric_column:
  alt_sinir=aykiri_degerler[column]["alt_sinir"]
  ust_sinir=aykiri_degerler[column]["ust_sinir"]
  aykiri_sayisi=((df[column]<alt_sinir)|(df[column]>ust_sinir)).sum()
  print(f"{column}:{aykiri_sayisi}")
```

```python
#46.Aykırı değer içeren kayıtları veri setinden temizleme
df_clean=df[~aykiri_maskesi].copy()
df_clean.shape
```

```python
#47.Temizlenmiş veri setini kaydet
df_clean.to_csv("e_ticaret_veri_seti_aykiri_degerler_temizlendi.csv")
```
