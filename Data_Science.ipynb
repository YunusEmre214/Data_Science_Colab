```python
#33.Şehir sütunundaki tutarsız değerleri düzeltme
def normalize(series):
  return(
      series.astype(str)
      .str.strip()
      .str.lower()
      .str.replace("i̇","i",regex=False)
      .str.replace("ı","i",regex=False)
      .str.replace("ş","s",regex=False)
      .str.replace("ç","c",regex=False)
      .str.replace("ö","o",regex=False)
      .str.replace("ğ","g",regex=False)
  )
```

```python
df["sehir"]=normalize(df["sehir"])
df["sehir"].unique()
```

```python
sehir_duzeltme_haritasi = {
    "istanbul": "İstanbul",
    "ankara": "Ankara",
    "izmir": "İzmir",
    "antalya": "Antalya",
    "konya": "Konya",
    "bursa": "Bursa",
    "adana": "Adana",
    "gaziantep": "Gaziantep",
    "mersin": "Mersin",
    "kayseri": "Kayseri",
    "eskisehir": "Eskişehir",
    "samsun": "Samsun",
    "trabzon": "Trabzon",
    "kocaeli": "Kocaeli"
}
df["sehir"]=df["sehir"].replace(sehir_duzeltme_haritasi)
df["sehir"].unique()
```

```python
#41.Sayısal sütunların belirlenmesi
numeric_column=df.select_dtypes(include=["int64","float64"]).columns.tolist()
numeric_column
```

```python
#42.Sayısal sütunların özet istatisliklerini incelemek
df[numeric_column].describe()
```

```python
"""
IQR=Q3-Q1=817-198=619 For birim_fiyat
Üst sınır=Q3+k*IQR
Alt sınır=Q1-k*IQR
k=1 olsun
Üst sınır=817+619=1436
ALt sınır=198-619=-421
Outlier=üst sınır ve alt sınır dışında kalan örnekler
"""
```

```python
#43.IQR ile aykırı değer inceleme
aykiri_degerler={}

for column in numeric_column:
  q1=df[column].quantile(0.25)
  q3=df[column].quantile(0.75)
  iqr=q3-q1
  ust_sinir=q3+1.5*iqr
  alt_sinir=q1-1.5*iqr

  aykiri_degerler[column]={
      "q1":q1,
      "q3":q3,
      "iqr":iqr,
      "ust_sinir":ust_sinir,
      "alt_sinir":alt_sinir
  }
```

```python
aykiri_degerler
```

```python
#44.Aykırı değer içeren kayıtları inceleyelim
aykiri_maskesi=pd.Series(False,index=df.index)
for column in numeric_column:
  alt_sinir=aykiri_degerler[column]["alt_sinir"]
  ust_sinir=aykiri_degerler[column]["ust_sinir"]
  sutun_maskesi=(df[column]<alt_sinir)|(df[column]>ust_sinir)
  aykiri_maskesi=aykiri_maskesi|sutun_maskesi
aykiri_kayıtlar=df[aykiri_maskesi]
aykiri_kayıtlar
```

```python
#45.Aykırı değer sayılarını sütun bazında incele
for column in numeric_column:
  alt_sinir=aykiri_degerler[column]["alt_sinir"]
  ust_sinir=aykiri_degerler[column]["ust_sinir"]
  aykiri_sayisi=((df[column]<alt_sinir)|(df[column]>ust_sinir)).sum()
  print(f"{column}:{aykiri_sayisi}")
```

```python
#46.Aykırı değer içeren kayıtları veri setinden temizleme
df_clean=df[~aykiri_maskesi].copy()
df_clean.shape
```

```python
#47.Temizlenmiş veri setini kaydet
df_clean.to_csv("e_ticaret_veri_seti_aykiri_degerler_temizlendi.csv")
```

```python
#78.Aylara göre sipariş yoğunluk
aylik_yogunluk=(
    df.groupby("siparis_ayi")
    .agg
    (
        siparis_sayisi=("siparis_id","count"),
        benzersiz_musteri_sayisi=("musteri_id","nunique")
    )
    .reset_index()
    .sort_values("siparis_ayi")
)
aylik_yogunluk
```

```python
plt.figure(figsize=(6,3))
plt.plot(aylik_yogunluk["siparis_ayi"],aylik_yogunluk["siparis_sayisi"],marker="o",color="red",label="siparis_sayisi")
plt.plot(aylik_yogunluk["siparis_ayi"],aylik_yogunluk["benzersiz_musteri_sayisi"],marker="o",color="blue",label="benzersiz_musteri_sayisi")
plt.title("Aylara Göre Sipariş Yoğunluğu")
plt.xlabel("Aylar")
plt.ylabel("Sipariş Sayısı/Benzersiz Müşteri Sayısı")
plt.legend()
plt.xticks(range(1,13))
plt.tight_layout()
plt.show()
```

```python
#79.Ayın ilk yarısı ve ikinci yarısında ki harcanan farkı inceleme
df["ay_donemi"]=df["siparis_gunu"].apply(
    lambda x: "Ayın ilk yarısı" if x<=15 else "Ayın ikinci yarısı"
)
df.head()
```

```python
ay_donemi_analizi=(
    df.groupby("ay_donemi")
    .agg(
        toplam_tutar=("toplam_tutar","sum"),
        benzersiz_musteri_sayisi=("musteri_id","nunique")
    )
    .reset_index()
)
ay_donemi_analizi
```

```python
plt.figure(figsize=(5,3))
sns.barplot(data=ay_donemi_analizi,x="ay_donemi",y="toplam_tutar")
plt.title("Ay Donemiye Göre Toplam Tutar")
plt.xlabel("Ay Donemi")
plt.ylabel("Toplam Tutar")
plt.show()
```

```python
#80.Hafta içi ve hafta sonu sipariş davranışı karşılaştırma
df["hafta_tipi"]=df["siparis_tarihi"].dt.dayofweek.apply(
    lambda x: "Hafta içi" if x<=4 else "Hafta sonu"
)
df.head()
```

```python
hafta_tipi_analizi=(
    df.groupby("hafta_tipi")
    .agg(
        toplam_tutar=("toplam_tutar","mean"),
        benzersiz_musteri_sayisi=("musteri_id","nunique")
    )
    .round(1)
    .reset_index()
)
hafta_tipi_analizi
```

```python
plt.figure(figsize=(5,3))
sns.barplot(data=hafta_tipi_analizi,x="hafta_tipi",y="toplam_tutar")
plt.title("Hafta Tipine Göre Ortalama Toplam Tutar")
plt.xlabel("Hafta Tipi")
plt.ylabel("Ortalama Toplam Tutar")
plt.show()
```

```python
#81.İki haftalık dönemlerde ortalama sipariş tutarı
df["iki_haftalik_donem"]=((df["hafta_numarasi"]-1)//2)+1
df.head()
```

```python
iki_haftalik_analiz=(
    df.groupby("iki_haftalik_donem")
    .agg(
        ortalama_toplam_tutar=("toplam_tutar","mean"),
        benzersiz_musteri_sayisi=("musteri_id","nunique")
    )
    .round(2)
    .reset_index()
    .sort_values("iki_haftalik_donem")
)
iki_haftalik_analiz
```

```python
plt.figure()
plt.plot(iki_haftalik_analiz["iki_haftalik_donem"],iki_haftalik_analiz["ortalama_toplam_tutar"],marker="o",color="red")
plt.title("İki Haftalık Dönemlerde Ortalama Sipariş Tutarı")
plt.xlabel("İki Haftalık Dönemler")
plt.ylabel("Ortalama Sipariş Tutarı")
plt.xticks(range(1,5))
plt.show()
```

```python
#85.Veri setini yükle
df=pd.read_csv("e_ticaret_veri_seti_aykiri_degerler_duzenlendi.csv")
df.head()
```

```python
#86.Hangi kategori hem yüksek gelir hem de müşteri memnuniyetini yüksek sağlıyor
kategori_raporu=(
    df.groupby("kategori")
    .agg(
        toplam_tutar=("toplam_tutar","sum"),
        ortalama_musteri_puani=("musteri_puani","mean")
    )
    .round(2)
)
kategori_raporu
```

```python
kategori_raporu=kategori_raporu.sort_values(by="toplam_tutar",ascending=False)
kategori_raporu
```

```python
#87.Kategoriye göre toplam gelir
plt.figure(figsize=(6,3))
sns.barplot(data=kategori_raporu,x=kategori_raporu.index,y="toplam_tutar")
plt.title("Kategoriye Göre Toplam Gelir")
plt.xlabel("Kategori")
plt.ylabel("Toplam Gelir")
plt.show()
```

```python
#88.Kategoriye göre ortalama musteri puanı
plt.figure(figsize=(6,3))
sns.barplot(data=kategori_raporu,x=kategori_raporu.index,y="ortalama_musteri_puani")
plt.title("Kategoriye Göre Toplam Gelir")
plt.xlabel("Kategori")
plt.ylabel("Müşteri Puanı")
plt.show()
```

```python
#89.Hangi ödeme türleri daha yüksek ortalama sipariş tutarı üretiyor ve müşteri puanı açısından performans gösteriyor
odeme_turu_raporu=(
    df.groupby("odeme_turu")
    .agg(
        ortalama_siparis_tutari=("toplam_tutar","mean"),
        ortalama_musteri_puani=("musteri_puani","mean")
    )
    .round(2)
    .sort_values(by="ortalama_siparis_tutari",ascending=False)
)
odeme_turu_raporu
```

```python
plt.figure(figsize=(8,3))
sns.barplot(data=odeme_turu_raporu,x="odeme_turu",y="ortalama_siparis_tutari")
plt.title("Ödeme Türüne Göre Ortalama Sipariş Tutarı")
plt.xlabel("Ödeme Türü")
plt.ylabel("Ortalama Sipariş Tutarı")
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.barplot(data=odeme_turu_raporu,x="odeme_turu",y="ortalama_musteri_puani")
plt.title("Ödeme Türüne Göre Ortalama Müşteri Puanı")
plt.xlabel("Ödeme Türü")
plt.ylabel("Ortalama Müşteri Puanı")
plt.show()
```
