```python
#10.sehir sütunundaki benzersiz değerleri inceleyin. İstanbul, istanbul, Istanbul gibi aynı anlama gelen farklı yazımları tek bir standart biçime dönüştürün. Temizlikten önce ve sonra benzersiz değer sayılarını karşılaştırın.
df_copy['sehir'].unique()
```

```python
#12.Aşağıdaki koşullara uyan mantıksal olarak hatalı kayıtları bulun: birim_fiyat <= 0, toplam_tutar <= 0, teslimat_gunu < 0, musteri_puani > 5. Bu kayıtların sayısını yazdırın ve veri setinden çıkarın.
hatali_kayitlar = df_copy[
    (df_copy["toplam_tutar"] <= 0) |
    (df_copy["birim_fiyat"] <= 0) |
    (df_copy["teslimat_gunu"] < 0) |
    (df_copy["musteri_puani"] > 5)
]
len(hatali_kayitlar)
```

```python
df_copy.shape
```

```python
hatali_kayitlar
```

```python
hatali_kosul = (
    (df_copy["toplam_tutar"] <= 0) |
    (df_copy["birim_fiyat"] <= 0) |
    (df_copy["teslimat_gunu"] < 0) |
    (df_copy["musteri_puani"] > 5)
)
df_copy = df_copy[~hatali_kosul]
df_copy.shape
```

```python
#Hatalı kayıtlar tespit edilerek ~ asterisk operatörü ile ilgili filtreleme yapılarak hatalı veriler temizlenmiştir.
```

```python
#13.birim_fiyat sütunu için IQR yöntemini kullanarak Q1, Q3, IQR, alt sınır ve üst sınır değerlerini hesaplayın. Sonra aykırı değer içeren kayıt sayısını yazdırın.
aykiri_sinirlar={}

q1 = df_copy["birim_fiyat"].quantile(0.25)
q3 = df_copy["birim_fiyat"].quantile(0.75)

iqr = q3 - q1
alt_sinir = q1 - 1.5 * iqr
ust_sinir = q3 + 1.5 * iqr

aykiri_sinirlar["birim_fiyat"] = {
      "q1": q1,
      "q3": q3,
      "iqr": iqr,
      "alt_sinir": alt_sinir,
      "ust_sinir": ust_sinir
  }
```

```python
aykiri_sinirlar
```

```python
aykiri_kosul = (df_copy['birim_fiyat'] < alt_sinir) | (df_copy['birim_fiyat'] > ust_sinir)
aykiri_sayisi = aykiri_kosul.sum()
aykiri_sayisi
```

```python
aykiri_maskesi = pd.Series(False, index = df_copy.index)
alt_sinir = aykiri_sinirlar["birim_fiyat"]["alt_sinir"]
ust_sinir = aykiri_sinirlar["birim_fiyat"]["ust_sinir"]

sutun_maskesi = (df_copy["birim_fiyat"] < alt_sinir) | (df_copy["birim_fiyat"] > ust_sinir)
aykiri_maskesi = aykiri_maskesi | sutun_maskesi
```

```python
aykiri_kayitlar = df_copy[aykiri_maskesi]
aykiri_kayitlar
```

```python
#IQR yöntemi kullanılarak birim fiyat için Q1,Q3 ve IQR değerleri hesaplandı ve aykırı değer kayıtları yazdırıldı.
```
